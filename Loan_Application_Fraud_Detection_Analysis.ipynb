```python

# ============================================================
# LOAN APPLICATION FRAUD DETECTION USING MACHINE LEARNING
# Advanced Academic Version
#
# Workflow:
# 1. Load and understand data
# 2. Data cleaning + leakage prevention
# 3. Feature engineering
# 4. EDA
# 5. Train/validation/test split
# 6. Preprocessing
# 7. 3 ML models:
#       - Logistic Regression
#       - Random Forest
#       - XGBoost
# 8. 3-fold stratified cross-validation
# 9. Threshold tuning on VALIDATION set
# 10. Test-set evaluation
# 11. ROC + Precision-Recall curves
# 12. Confusion matrices
# 13. Business-cost analysis
# 14. SHAP explainability for XGBoost
# 15. Final model comparison and conclusions
#
# IMPORTANT:
# - fraud_type and loan_status are excluded to prevent target leakage.
# - IDs and raw residential address are excluded.
# - The test set is never used for threshold selection or model tuning.
# ============================================================

# ============================================================
# 1. INSTALL LIBRARIES
# ============================================================

!pip -q install xgboost shap

# ============================================================
# 2. IMPORT LIBRARIES
# ============================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    precision_recall_curve
)

import shap

from IPython.display import display

sns.set_theme(style="whitegrid")

RANDOM_STATE = 42


# ============================================================
# 3. LOAD DATASET
# ============================================================

file_path = "/content/loan_applications.csv"

if not os.path.exists(file_path):
    from google.colab import files

    print("Please upload loan_applications.csv")
    uploaded = files.upload()
    file_path = "/content/" + list(uploaded.keys())[0]

df = pd.read_csv(file_path)

print("=" * 80)
print("DATASET LOADED")
print("=" * 80)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")

display(df.head())


# ============================================================
# 4. INITIAL DATA UNDERSTANDING
# ============================================================

print("\n" + "=" * 80)
print("COLUMN NAMES")
print("=" * 80)

print(df.columns.tolist())

print("\n" + "=" * 80)
print("DATA TYPES")
print("=" * 80)

print(df.dtypes)

print("\n" + "=" * 80)
print("DATASET INFO")
print("=" * 80)

df.info()

print("\n" + "=" * 80)
print("SUMMARY STATISTICS")
print("=" * 80)

display(df.describe(include="all").T)


# ============================================================
# 5. DATA QUALITY CHECK
# ============================================================

print("\n" + "=" * 80)
print("MISSING VALUES")
print("=" * 80)

missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_table = pd.DataFrame({
    "Missing Count": missing,
    "Missing %": missing_pct
})

display(
    missing_table[missing_table["Missing Count"] > 0]
    .sort_values("Missing Count", ascending=False)
)

print("\n" + "=" * 80)
print("DUPLICATES")
print("=" * 80)

print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate application IDs:", df["application_id"].duplicated().sum())
print("Duplicate customer IDs:", df["customer_id"].duplicated().sum())


# ============================================================
# 6. TARGET ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("TARGET DISTRIBUTION")
print("=" * 80)

target_counts = df["fraud_flag"].value_counts().sort_index()
target_pct = df["fraud_flag"].value_counts(normalize=True).sort_index() * 100

target_summary = pd.DataFrame({
    "Count": target_counts,
    "Percentage": target_pct.round(2)
})

display(target_summary)

fraud_rate = df["fraud_flag"].mean()

print(f"\nOverall fraud rate: {fraud_rate * 100:.2f}%")

plt.figure(figsize=(7, 5))
sns.countplot(data=df, x="fraud_flag")
plt.title("Fraud vs Legitimate Loan Applications")
plt.xlabel("Fraud Flag (0 = Legitimate, 1 = Fraud)")
plt.ylabel("Number of Applications")
plt.show()

plt.figure(figsize=(6, 6))
plt.pie(
    target_counts.values,
    labels=["Legitimate", "Fraudulent"],
    autopct="%1.2f%%",
    startangle=90
)
plt.title("Loan Application Fraud Distribution")
plt.show()


# ============================================================
# 7. FRAUD TYPE ANALYSIS
# ============================================================
# Used ONLY for descriptive EDA.
# It is NOT used as a model feature.

print("\n" + "=" * 80)
print("FRAUD TYPES")
print("=" * 80)

fraud_type_counts = (
    df.loc[df["fraud_flag"] == 1, "fraud_type"]
    .value_counts()
)

display(fraud_type_counts)

plt.figure(figsize=(9, 5))
sns.countplot(
    data=df[df["fraud_flag"] == 1],
    y="fraud_type",
    order=fraud_type_counts.index
)
plt.title("Types of Fraudulent Loan Applications")
plt.xlabel("Number of Fraudulent Applications")
plt.ylabel("Fraud Type")
plt.show()


# ============================================================
# 8. DATA CLEANING
# ============================================================

data = df.copy()

# Remove exact duplicates
data = data.drop_duplicates().copy()

# Convert application date
data["application_date"] = pd.to_datetime(
    data["application_date"],
    errors="coerce"
)


# ============================================================
# 9. FEATURE ENGINEERING
# ============================================================

# Calendar features
data["application_year"] = data["application_date"].dt.year
data["application_month"] = data["application_date"].dt.month
data["application_dayofweek"] = data["application_date"].dt.dayofweek
data["is_weekend_application"] = (
    data["application_dayofweek"] >= 5
).astype(int)

# Financial ratios
data["loan_to_annual_income"] = (
    data["loan_amount_requested"] /
    (data["monthly_income"] * 12 + 1)
)

data["emi_to_income_ratio"] = (
    data["existing_emis_monthly"] /
    (data["monthly_income"] + 1)
)

data["estimated_monthly_disposable_income"] = (
    data["monthly_income"] -
    data["existing_emis_monthly"]
)

data["loan_amount_per_tenure_month"] = (
    data["loan_amount_requested"] /
    (data["loan_tenure_months"] + 1)
)


# ============================================================
# 10. TARGET LEAKAGE PREVENTION
# ============================================================
#
# DO NOT use:
#
# fraud_type:
#   Directly describes the type of fraud.
#
# loan_status:
#   Contains information such as "Fraudulent - Detected"
#   and "Fraudulent - Undetected".
#
# Also remove identifiers / raw address:
# application_id
# customer_id
# residential_address
#
# This ensures the model learns from information that could
# reasonably be available when screening a new application.
# ============================================================

columns_to_remove = [
    "application_id",
    "customer_id",
    "application_date",
    "residential_address",
    "loan_status",
    "fraud_type"
]

data_model = data.drop(
    columns=columns_to_remove,
    errors="ignore"
).copy()


# ============================================================
# 11. CLEANED DATA CHECK
# ============================================================

print("\n" + "=" * 80)
print("CLEANED MODELING DATA")
print("=" * 80)

print("Rows:", data_model.shape[0])
print("Columns:", data_model.shape[1])

display(data_model.head())

print("\nRemaining missing values:")
display(
    data_model.isnull().sum()
    .sort_values(ascending=False)
    .head(20)
)


# ============================================================
# 12. EXPLORATORY DATA ANALYSIS
# ============================================================

numeric_eda = [
    "loan_amount_requested",
    "loan_tenure_months",
    "interest_rate_offered",
    "monthly_income",
    "cibil_score",
    "existing_emis_monthly",
    "debt_to_income_ratio",
    "applicant_age",
    "number_of_dependents"
]

# Fraud vs legitimate mean values
fraud_numeric_summary = (
    data.groupby("fraud_flag")[numeric_eda]
    .mean()
    .T
)

fraud_numeric_summary.columns = [
    "Legitimate",
    "Fraudulent"
]

print("\n" + "=" * 80)
print("AVERAGE NUMERICAL VARIABLES: FRAUD VS LEGITIMATE")
print("=" * 80)

display(fraud_numeric_summary.round(2))


# Boxplots
for col in numeric_eda:

    plt.figure(figsize=(8, 5))

    sns.boxplot(
        data=data,
        x="fraud_flag",
        y=col
    )

    plt.title(
        f"{col.replace('_', ' ').title()} by Fraud Status"
    )

    plt.xlabel("Fraud Flag")
    plt.ylabel(col.replace("_", " ").title())

    plt.show()


# ============================================================
# 13. CATEGORICAL EDA
# ============================================================

categorical_eda = [
    "loan_type",
    "purpose_of_loan",
    "employment_status",
    "property_ownership_status",
    "gender"
]

for col in categorical_eda:

    print("\n" + "=" * 80)
    print(f"FRAUD RATE BY {col.upper()}")
    print("=" * 80)

    fraud_rate_table = (
        data.groupby(col)["fraud_flag"]
        .agg(["count", "sum", "mean"])
        .sort_values("mean", ascending=False)
    )

    fraud_rate_table["fraud_rate_percent"] = (
        fraud_rate_table["mean"] * 100
    )

    display(fraud_rate_table.round(4))

    plot_data = (
        fraud_rate_table["fraud_rate_percent"]
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(10, 5))

    sns.barplot(
        x=plot_data.values,
        y=plot_data.index
    )

    plt.title(
        f"Fraud Rate by {col.replace('_', ' ').title()}"
    )

    plt.xlabel("Fraud Rate (%)")
    plt.ylabel(col.replace("_", " ").title())

    plt.show()


# ============================================================
# 14. CORRELATION ANALYSIS
# ============================================================

numeric_corr = data_model.select_dtypes(
    include=np.number
)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_corr.corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Matrix of Numerical Variables")
plt.show()


# ============================================================
# 15. PREPARE FEATURES AND TARGET
# ============================================================

X = data_model.drop(
    columns=["fraud_flag"]
)

y = data_model["fraud_flag"]


# ============================================================
# 16. TRAIN / VALIDATION / TEST SPLIT
# ============================================================
#
# 60% Training
# 20% Validation
# 20% Test
#
# Training:
#   Model learning + cross-validation
#
# Validation:
#   Threshold selection + business-cost optimization
#
# Test:
#   Final unbiased performance evaluation
# ============================================================

X_temp, X_test, y_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_temp,
    y_temp,
    test_size=0.25,
    random_state=RANDOM_STATE,
    stratify=y_temp
)

print("\n" + "=" * 80)
print("DATA SPLIT")
print("=" * 80)

print("Training samples  :", len(X_train))
print("Validation samples:", len(X_val))
print("Testing samples   :", len(X_test))

print("\nFraud rates:")
print("Train:", round(y_train.mean() * 100, 2), "%")
print("Validation:", round(y_val.mean() * 100, 2), "%")
print("Test:", round(y_test.mean() * 100, 2), "%")


# ============================================================
# 17. IDENTIFY FEATURE TYPES
# ============================================================

numeric_features = X_train.select_dtypes(
    include=["int64", "float64", "int32", "float32"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("\nNumerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)


# ============================================================
# 18. PREPROCESSING PIPELINE
# ============================================================

numeric_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="median")
        ),
        (
            "scaler",
            StandardScaler()
        )
    ]
)

categorical_transformer = Pipeline(
    steps=[
        (
            "imputer",
            SimpleImputer(strategy="most_frequent")
        ),
        (
            "onehot",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            )
        )
    ]
)

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            numeric_transformer,
            numeric_features
        ),
        (
            "cat",
            categorical_transformer,
            categorical_features
        )
    ]
)


# ============================================================
# 19. DEFINE MODELS
# ============================================================

# Model 1: Logistic Regression
logistic_model = LogisticRegression(
    class_weight="balanced",
    max_iter=2000,
    random_state=RANDOM_STATE
)


# Model 2: Random Forest
random_forest_model = RandomForestClassifier(
    n_estimators=250,
    max_depth=12,
    min_samples_leaf=3,
    class_weight="balanced_subsample",
    random_state=RANDOM_STATE,
    n_jobs=-1
)


# Model 3: XGBoost
negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print(
    "\nXGBoost scale_pos_weight:",
    round(scale_pos_weight, 2)
)

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.80,
    colsample_bytree=0.80,
    min_child_weight=3,
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=scale_pos_weight,
    random_state=RANDOM_STATE,
    n_jobs=-1
)


# ============================================================
# 20. CREATE COMPLETE PIPELINES
# ============================================================

models = {

    "Logistic Regression": Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", logistic_model)
        ]
    ),

    "Random Forest": Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", random_forest_model)
        ]
    ),

    "XGBoost": Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            ("model", xgb_model)
        ]
    )
}


# ============================================================
# 21. 3-FOLD STRATIFIED CROSS-VALIDATION
# ============================================================
#
# Cross-validation is performed ONLY on the training data.
# The validation and test sets remain untouched.
#
# Metrics:
# - F1
# - ROC-AUC
# - PR-AUC
#
# PR-AUC is especially useful for this imbalanced fraud problem.
# ============================================================

cv = StratifiedKFold(
    n_splits=3,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_results = []

print("\n" + "=" * 80)
print("3-FOLD STRATIFIED CROSS-VALIDATION")
print("=" * 80)

for model_name, model in models.items():

    print(f"\nRunning CV for {model_name}...")

    scores = cross_validate(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring={
            "f1": "f1",
            "roc_auc": "roc_auc",
            "pr_auc": "average_precision"
        },
        n_jobs=1
    )

    cv_results.append({
        "Model": model_name,
        "CV F1 Mean": scores["test_f1"].mean(),
        "CV F1 Std": scores["test_f1"].std(),
        "CV ROC-AUC Mean": scores["test_roc_auc"].mean(),
        "CV PR-AUC Mean": scores["test_pr_auc"].mean()
    })

cv_results_df = pd.DataFrame(cv_results)

print("\nCross-validation results:")
display(
    cv_results_df.style.format({
        "CV F1 Mean": "{:.4f}",
        "CV F1 Std": "{:.4f}",
        "CV ROC-AUC Mean": "{:.4f}",
        "CV PR-AUC Mean": "{:.4f}"
    })
)


# ============================================================
# 22. TRAIN MODELS ON TRAINING SET
# ============================================================

fitted_models = {}

validation_probabilities = {}
test_probabilities = {}

print("\n" + "=" * 80)
print("TRAINING FINAL MODEL VERSIONS")
print("=" * 80)

for model_name, model in models.items():

    print(f"Training {model_name}...")

    model.fit(X_train, y_train)

    fitted_models[model_name] = model

    validation_probabilities[model_name] = (
        model.predict_proba(X_val)[:, 1]
    )

    test_probabilities[model_name] = (
        model.predict_proba(X_test)[:, 1]
    )


# ============================================================
# 23. THRESHOLD TUNING
# ============================================================
#
# Default classification threshold = 0.50.
#
# In fraud detection, 0.50 is NOT necessarily optimal.
#
# We search thresholds on the VALIDATION set only.
#
# The threshold is selected using F1 score.
# ============================================================

def evaluate_at_threshold(y_true, probabilities, threshold):

    predictions = (
        probabilities >= threshold
    ).astype(int)

    return {
        "Threshold": threshold,
        "Accuracy": accuracy_score(y_true, predictions),
        "Precision": precision_score(
            y_true,
            predictions,
            zero_division=0
        ),
        "Recall": recall_score(
            y_true,
            predictions,
            zero_division=0
        ),
        "F1": f1_score(
            y_true,
            predictions,
            zero_division=0
        )
    }


thresholds = np.arange(
    0.05,
    0.96,
    0.01
)

optimal_thresholds = {}

threshold_tables = {}

print("\n" + "=" * 80)
print("VALIDATION-SET THRESHOLD OPTIMIZATION")
print("=" * 80)

for model_name, probabilities in validation_probabilities.items():

    threshold_results = []

    for threshold in thresholds:

        threshold_results.append(
            evaluate_at_threshold(
                y_val,
                probabilities,
                threshold
            )
        )

    threshold_df = pd.DataFrame(
        threshold_results
    )

    best_row = threshold_df.loc[
        threshold_df["F1"].idxmax()
    ]

    optimal_thresholds[model_name] = (
        best_row["Threshold"]
    )

    threshold_tables[model_name] = threshold_df

    print(
        f"{model_name}: "
        f"Best threshold = "
        f"{best_row['Threshold']:.2f}, "
        f"Validation F1 = "
        f"{best_row['F1']:.4f}"
    )


# ============================================================
# 24. PLOT VALIDATION F1 VS THRESHOLD
# ============================================================

plt.figure(figsize=(10, 6))

for model_name, threshold_df in threshold_tables.items():

    plt.plot(
        threshold_df["Threshold"],
        threshold_df["F1"],
        label=model_name
    )

plt.title("Validation F1 Score Across Classification Thresholds")
plt.xlabel("Fraud Probability Threshold")
plt.ylabel("F1 Score")
plt.legend()
plt.show()


# ============================================================
# 25. BUSINESS COST ANALYSIS
# ============================================================
#
# Illustrative business assumption:
#
# False Negative (missed fraud) = cost 10
# False Positive (legitimate application flagged) = cost 1
#
# This is NOT a real bank cost estimate.
# It is an academic assumption used to demonstrate
# cost-sensitive threshold selection.
# ============================================================

FALSE_NEGATIVE_COST = 10
FALSE_POSITIVE_COST = 1

def calculate_business_cost(y_true, probabilities, threshold):

    predictions = (
        probabilities >= threshold
    ).astype(int)

    cm = confusion_matrix(
        y_true,
        predictions
    )

    tn, fp, fn, tp = cm.ravel()

    total_cost = (
        fn * FALSE_NEGATIVE_COST +
        fp * FALSE_POSITIVE_COST
    )

    return {
        "Threshold": threshold,
        "False Positives": fp,
        "False Negatives": fn,
        "Business Cost": total_cost
    }


cost_optimal_thresholds = {}

print("\n" + "=" * 80)
print("BUSINESS-COST THRESHOLD ANALYSIS")
print("=" * 80)

for model_name, probabilities in validation_probabilities.items():

    cost_results = []

    for threshold in thresholds:

        cost_results.append(
            calculate_business_cost(
                y_val,
                probabilities,
                threshold
            )
        )

    cost_df = pd.DataFrame(cost_results)

    best_cost_row = cost_df.loc[
        cost_df["Business Cost"].idxmin()
    ]

    cost_optimal_thresholds[model_name] = (
        best_cost_row["Threshold"]
    )

    print(
        f"\n{model_name}"
    )

    print(
        f"Cost-optimal threshold: "
        f"{best_cost_row['Threshold']:.2f}"
    )

    print(
        f"False positives: "
        f"{int(best_cost_row['False Positives'])}"
    )

    print(
        f"False negatives: "
        f"{int(best_cost_row['False Negatives'])}"
    )

    print(
        f"Illustrative business cost: "
        f"{int(best_cost_row['Business Cost'])}"
    )


# ============================================================
# 26. FINAL TEST-SET EVALUATION
# ============================================================
#
# Test set is evaluated ONLY AFTER:
# - model training
# - cross-validation
# - threshold selection
#
# This prevents test-set leakage.
# ============================================================

final_results = []

test_predictions = {}

print("\n" + "=" * 80)
print("FINAL TEST-SET MODEL EVALUATION")
print("=" * 80)

for model_name in models.keys():

    probabilities = test_probabilities[model_name]

    # Use F1-optimized validation threshold
    threshold = optimal_thresholds[model_name]

    predictions = (
        probabilities >= threshold
    ).astype(int)

    test_predictions[model_name] = predictions

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        probabilities
    )

    pr_auc = average_precision_score(
        y_test,
        probabilities
    )

    cm = confusion_matrix(
        y_test,
        predictions
    )

    tn, fp, fn, tp = cm.ravel()

    business_cost = (
        fn * FALSE_NEGATIVE_COST +
        fp * FALSE_POSITIVE_COST
    )

    final_results.append({
        "Model": model_name,
        "Threshold": threshold,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1,
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
        "False Positives": fp,
        "False Negatives": fn,
        "Business Cost": business_cost
    })

    print("\n" + "-" * 70)
    print(model_name)
    print("-" * 70)

    print(
        classification_report(
            y_test,
            predictions,
            target_names=[
                "Legitimate",
                "Fraudulent"
            ],
            zero_division=0
        )
    )


final_results_df = pd.DataFrame(
    final_results
).sort_values(
    by="F1 Score",
    ascending=False
)

print("\n" + "=" * 80)
print("FINAL MODEL COMPARISON")
print("=" * 80)

display(
    final_results_df.style.format({
        "Threshold": "{:.2f}",
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1 Score": "{:.4f}",
        "ROC-AUC": "{:.4f}",
        "PR-AUC": "{:.4f}"
    })
)


# ============================================================
# 27. MODEL PERFORMANCE VISUALIZATION
# ============================================================

metrics_to_plot = [
    "Precision",
    "Recall",
    "F1 Score",
    "ROC-AUC",
    "PR-AUC"
]

plot_df = final_results_df.melt(
    id_vars="Model",
    value_vars=metrics_to_plot,
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(13, 6))

sns.barplot(
    data=plot_df,
    x="Metric",
    y="Score",
    hue="Model"
)

plt.ylim(0, 1)

plt.title(
    "Final Test-Set Model Performance Comparison"
)

plt.xlabel("Metric")
plt.ylabel("Score")

plt.legend(
    title="Model",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.show()


# ============================================================
# 28. CONFUSION MATRICES
# ============================================================

for model_name in models.keys():

    cm = confusion_matrix(
        y_test,
        test_predictions[model_name]
    )

    plt.figure(figsize=(5, 4))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=[
            "Legitimate",
            "Fraudulent"
        ],
        yticklabels=[
            "Legitimate",
            "Fraudulent"
        ]
    )

    plt.title(
        f"Confusion Matrix - {model_name}"
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")

    plt.show()


# ============================================================
# 29. ROC CURVE
# ============================================================

plt.figure(figsize=(9, 7))

for model_name in models.keys():

    fpr, tpr, _ = roc_curve(
        y_test,
        test_probabilities[model_name]
    )

    auc_score = roc_auc_score(
        y_test,
        test_probabilities[model_name]
    )

    plt.plot(
        fpr,
        tpr,
        label=f"{model_name} (AUC = {auc_score:.3f})"
    )

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.title("ROC Curves - Fraud Detection Models")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.show()


# ============================================================
# 30. PRECISION-RECALL CURVE
# ============================================================

plt.figure(figsize=(9, 7))

for model_name in models.keys():

    precision_curve, recall_curve, _ = (
        precision_recall_curve(
            y_test,
            test_probabilities[model_name]
        )
    )

    ap_score = average_precision_score(
        y_test,
        test_probabilities[model_name]
    )

    plt.plot(
        recall_curve,
        precision_curve,
        label=f"{model_name} (PR-AUC = {ap_score:.3f})"
    )

plt.title(
    "Precision-Recall Curves - Fraud Detection"
)

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.legend()
plt.show()


# ============================================================
# 31. SELECT BEST MODEL
# ============================================================
#
# Primary criterion: F1 score.
#
# Why?
# Fraud detection needs a balance between:
# - catching fraud (recall)
# - avoiding too many false alarms (precision)
#
# PR-AUC is also reported because the fraud class is rare.
# ============================================================

best_model_name = (
    final_results_df
    .sort_values(
        by=["F1 Score", "PR-AUC"],
        ascending=False
    )
    .iloc[0]["Model"]
)

best_model = fitted_models[best_model_name]

best_threshold = optimal_thresholds[
    best_model_name
]

print("\n" + "=" * 80)
print("BEST MODEL")
print("=" * 80)

print("Best model:", best_model_name)
print("Selected probability threshold:", round(best_threshold, 2))


# ============================================================
# 32. SHAP EXPLAINABILITY
# ============================================================
#
# SHAP explains how features contribute to the model's
# prediction.
#
# We use SHAP on the XGBoost model because tree-based SHAP
# provides efficient and useful explanations for boosted trees.
#
# If XGBoost is the best model, SHAP is applied directly to it.
# If another model wins, we still show XGBoost SHAP as an
# advanced model interpretation section.
# ============================================================

print("\n" + "=" * 80)
print("SHAP MODEL EXPLAINABILITY")
print("=" * 80)

xgb_pipeline_fitted = fitted_models["XGBoost"]

xgb_preprocessor = (
    xgb_pipeline_fitted
    .named_steps["preprocessor"]
)

xgb_classifier = (
    xgb_pipeline_fitted
    .named_steps["model"]
)

# Transform a sample of the test data
SHAP_SAMPLE_SIZE = min(1000, len(X_test))

X_shap_raw = X_test.sample(
    n=SHAP_SAMPLE_SIZE,
    random_state=RANDOM_STATE
)

X_shap_transformed = (
    xgb_preprocessor.transform(
        X_shap_raw
    )
)

feature_names = (
    xgb_preprocessor
    .get_feature_names_out()
)

feature_names_clean = [
    name.replace("num__", "")
        .replace("cat__", "")
    for name in feature_names
]

# Convert sparse matrix if necessary
if hasattr(X_shap_transformed, "toarray"):
    X_shap_for_model = X_shap_transformed.toarray()
else:
    X_shap_for_model = X_shap_transformed

X_shap_df = pd.DataFrame(
    X_shap_for_model,
    columns=feature_names_clean,
    index=X_shap_raw.index
)

# SHAP TreeExplainer
explainer = shap.TreeExplainer(
    xgb_classifier
)

shap_values = explainer.shap_values(
    X_shap_df
)


# ============================================================
# 33. SHAP SUMMARY PLOT
# ============================================================

plt.figure()

shap.summary_plot(
    shap_values,
    X_shap_df,
    plot_type="bar",
    max_display=20,
    show=False
)

plt.title(
    "Top Features Influencing XGBoost Fraud Predictions"
)

plt.tight_layout()
plt.show()


# ============================================================
# 34. SHAP DETAILED SUMMARY
# ============================================================

shap.summary_plot(
    shap_values,
    X_shap_df,
    max_display=20
)


# ============================================================
# 35. XGBOOST FEATURE IMPORTANCE
# ============================================================

xgb_importance = pd.DataFrame({
    "Feature": feature_names_clean,
    "Importance": xgb_classifier.feature_importances_
})

xgb_importance = (
    xgb_importance
    .sort_values(
        "Importance",
        ascending=False
    )
    .head(20)
)

print("\n" + "=" * 80)
print("TOP 20 XGBOOST FEATURES")
print("=" * 80)

display(xgb_importance)


plt.figure(figsize=(10, 8))

sns.barplot(
    data=xgb_importance,
    x="Importance",
    y="Feature"
)

plt.title("Top XGBoost Feature Importances")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.show()


# ============================================================
# 36. FRAUD PATTERN ANALYSIS
# ============================================================

print("\n" + "=" * 80)
print("KEY FRAUD PATTERNS")
print("=" * 80)

overall_fraud_rate = (
    data["fraud_flag"].mean() * 100
)

print(
    f"Overall fraud rate: "
    f"{overall_fraud_rate:.2f}%"
)

for col in categorical_eda:

    category_rates = (
        data.groupby(col)["fraud_flag"]
        .mean()
        .sort_values(ascending=False)
    )

    highest_category = category_rates.index[0]
    highest_rate = category_rates.iloc[0] * 100

    lowest_category = category_rates.index[-1]
    lowest_rate = category_rates.iloc[-1] * 100

    print(
        f"\n{col.replace('_', ' ').title()}:"
    )

    print(
        f"Highest observed fraud rate: "
        f"{highest_category} "
        f"({highest_rate:.2f}%)"
    )

    print(
        f"Lowest observed fraud rate: "
        f"{lowest_category} "
        f"({lowest_rate:.2f}%)"
    )


# ============================================================
# 37. FRAUD VS LEGITIMATE NUMERICAL COMPARISON
# ============================================================

comparison = (
    data.groupby("fraud_flag")[numeric_eda]
    .agg(["mean", "median"])
)

print("\n" + "=" * 80)
print("FRAUD VS LEGITIMATE NUMERICAL COMPARISON")
print("=" * 80)

display(comparison.round(2))


# ============================================================
# 38. BUSINESS INTERPRETATION
# ============================================================

best_row = final_results_df.iloc[0]

print("\n" + "=" * 80)
print("BUSINESS INTERPRETATION")
print("=" * 80)

print(
    f"""
The dataset contains approximately {len(data):,} loan applications,
of which {int(data['fraud_flag'].sum()):,} are fraudulent.

Fraud represents approximately
{data['fraud_flag'].mean() * 100:.2f}% of applications.

Because fraud is relatively rare, accuracy alone is not an
appropriate metric for judging the model.

The analysis therefore emphasizes:
- Precision
- Recall
- F1 Score
- ROC-AUC
- PR-AUC
- Confusion Matrix
- Business cost of false positives and false negatives

The best model based on the selected evaluation framework is:
{best_model_name}

Its validation-selected probability threshold is:
{best_threshold:.2f}

A practical banking implementation should NOT automatically
reject every application flagged by the model.

Instead, the model can act as a risk-screening layer:

LOW RISK
    -> Normal processing

MEDIUM RISK
    -> Additional verification

HIGH RISK
    -> Enhanced due diligence / manual review

This reduces the risk of both:
1. approving fraudulent applications, and
2. unnecessarily rejecting legitimate customers.
"""
)


# ============================================================
# 39. SAVE RESULTS
# ============================================================

final_results_df.to_csv(
    "/content/final_model_performance.csv",
    index=False
)

cv_results_df.to_csv(
    "/content/cross_validation_results.csv",
    index=False
)

xgb_importance.to_csv(
    "/content/xgboost_feature_importance.csv",
    index=False
)

print("\n" + "=" * 80)
print("FILES SAVED")
print("=" * 80)

print("/content/final_model_performance.csv")
print("/content/cross_validation_results.csv")
print("/content/xgboost_feature_importance.csv")


# ============================================================
# 40. FINAL PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 80)
print("FINAL PROJECT SUMMARY")
print("=" * 80)

print(
    f"""
PROJECT:
Loan Application Fraud Detection Using Machine Learning

DATASET:
{len(data):,} loan applications

LEGITIMATE:
{int((data['fraud_flag'] == 0).sum()):,}

FRAUDULENT:
{int((data['fraud_flag'] == 1).sum()):,}

FRAUD RATE:
{data['fraud_flag'].mean() * 100:.2f}%

MODELS:
1. Logistic Regression
2. Random Forest
3. XGBoost

VALIDATION:
3-fold Stratified Cross-Validation

THRESHOLD:
Optimized on validation data rather than blindly using 0.50

EVALUATION:
Precision
Recall
F1 Score
ROC-AUC
PR-AUC
Confusion Matrix
Business Cost

EXPLAINABILITY:
SHAP + XGBoost Feature Importance

BEST MODEL:
{best_model_name}

IMPORTANT:
The test set was kept separate and was not used for
threshold tuning or model training.
"""
)

print("\nPROJECT EXECUTION COMPLETE.")
```
